### **Exercise 2: Arithmetic with Arrays (3 min)**

Given the 1D array `arr = np.array([1, 2, 3, 4])`, perform the following:

1. Multiply every element by 3.
2. Add another array `np.array([10, 20, 30, 40])` to `arr`.
3. Compute the square of each element in `arr`.

````python
# Type your solution here:
arr = np.array([1, 2, 3, 4])

# Multiplying each element by 3
print(arr * 3)

#Add another array 
print(arr + np.array([10, 20, 30, 40]))

#Square of each element in the array
print(arr ** 2)
````

### **Exercise 3: Basic Indexing and Slicing (3 min)**

Given the 2D array:

```python
arr2d = np.array([[10, 20, 30],
                  [40, 50, 60],
                  [70, 80, 90]])
```

1. Select the element `50`.
2. Select the **first row**.
3. Select the **last column**.
4. Select a **2x2 subarray** containing the top-left four elements.

````python
# Type your solution here:
arr2d = np.array([[10, 20, 30], [40, 50, 60],[70, 80, 90]])
print(arr2d)

# Selecting the element 50
print(arr2d[1][1])

# Select the first row
print(arr2d[0])

#Select the last column
print(arr2d[2])

#Select 2*2 subarry
print(arr2d[:2, 1:])
````

### **Exercise 4: Boolean Indexing (2 min)**

### **Exercise 6: Transposing and Swapping Axes (3 min)**

Given the 2D array:

```python
arr2d = np.array([[1, 2, 3],
                  [4, 5, 6]])
```

1. Transpose the array.
2. Swap the two axes of `arr2d` (note: `arr2d` has shape (2, 3)).
3. Print the shape of the original and transposed arrays.


````python
# Type your solution here:
arr2d2 = np.array([[1, 2, 3], [4, 5, 6]])

#Transponse Array
transpose = arr2d2.T
print(transpose)

#Swapp axes 
print(arr2d2.swapaxes(0,1))

#Printing the shape of original and transposed array
print("Original  array shape: " , arr2d2.shape)
print("Transponse  array shape: " , transpose.shape)
````
